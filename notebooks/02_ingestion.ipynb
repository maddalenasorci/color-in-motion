```python
import pandas as pd

films = pd.read_csv("../data/processed/film_with_trailers.csv")
print("Film caricati:", len(films))
films.head()
```

```python
import os
import yt_dlp

# Cartella dove salviamo i trailer
os.makedirs("../data/raw/trailers", exist_ok=True)

# Prendiamo i primi 3 film per la prova
prova = films.head(3)

for i, row in prova.iterrows():
    tconst = row["tconst"]
    link = row["trailer_link"]
    output = f"../data/raw/trailers/{tconst}.mp4"

    print(f"Scarico {tconst}...")

    opzioni = {
        "format": "best[height<=360]",       # qualità massima entro 360p (leggera)
        "outtmpl": output,                    # dove salvare e con che nome
        "extractor_args": {                   # il trucco YouTube della Fase 0:
            "youtube": {"player_client": ["android"]}   # fingiti l'app Android
        },
        "quiet": True,                        # meno messaggi a schermo
    }

    with yt_dlp.YoutubeDL(opzioni) as ydl:
        ydl.download([link])

    print(f"  fatto: {tconst}.mp4")

print("Prova finita!")
```

```python
import time

# Cartella trailer (gia esiste, ma per sicurezza)
os.makedirs("../data/raw/trailers", exist_ok=True)

falliti = []   # qui teniamo i film che non si scaricano

for i, row in films.iterrows():
    tconst = row["tconst"]
    link = row["trailer_link"]
    output = f"../data/raw/trailers/{tconst}.mp4"

    # CACHE: se il file esiste gia, salta
    if os.path.exists(output):
        continue

    opzioni = {
        "format": "best[height<=360]",
        "outtmpl": output,
        "extractor_args": {"youtube": {"player_client": ["android"]}},
        "quiet": True,
    }

    # GESTIONE ERRORI: se fallisce, registra e vai avanti
    try:
        with yt_dlp.YoutubeDL(opzioni) as ydl:
            ydl.download([link])
    except Exception as e:
        falliti.append(tconst)
        print(f"  FALLITO {tconst}")

    # ogni 25 film stampiamo a che punto siamo
    if i % 25 == 0:
        print(f"Fatti {i} film...")

    time.sleep(0.5)   # pausa tra un download e l'altro

print("Download finito!")
print("Trailer falliti:", len(falliti))
```

```python
import time

# Mescoliamo il campione: cosi i generi si alternano
films_shuffled = films.sample(frac=1, random_state=42).reset_index(drop=True)

falliti = []
scaricati_ora = 0

for i, row in films_shuffled.iterrows():
    tconst = row["tconst"]
    link = row["trailer_link"]
    output = f"../data/raw/trailers/{tconst}.mp4"

    if os.path.exists(output):
        continue

    opzioni = {
        "format": "best[height<=480]/best",
        "outtmpl": output,
        "extractor_args": {"youtube": {"player_client": ["android"]}},
        "quiet": True,
    }

    try:
        with yt_dlp.YoutubeDL(opzioni) as ydl:
            ydl.download([link])
        scaricati_ora += 1
    except Exception as e:
        falliti.append(tconst)

    # ci fermiamo dopo 40 nuovi download per non farci bloccare
    if scaricati_ora >= 40:
        print("Lotto di 40 completato, mi fermo.")
        break

    time.sleep(2)   # pausa piu lunga: 2 secondi

print("Scaricati in questo lotto:", scaricati_ora)
print("Falliti:", len(falliti))
```

```python
import time

# Generi che ci servono ancora (Action ne abbiamo gia troppi)
generi_mancanti = ["Drama", "Comedy", "Crime", "Horror"]

# Prendiamo solo i film di questi generi, mescolati
da_scaricare = films[films["genere_principale"].isin(generi_mancanti)]
da_scaricare = da_scaricare.sample(frac=1, random_state=1).reset_index(drop=True)

scaricati_ora = 0
for i, row in da_scaricare.iterrows():
    tconst = row["tconst"]
    output = f"../data/raw/trailers/{tconst}.mp4"

    if os.path.exists(output):
        continue

    opzioni = {
        "format": "best[height<=480]/best",
        "outtmpl": output,
        "extractor_args": {"youtube": {"player_client": ["android"]}},
        "quiet": True,
    }

    try:
        with yt_dlp.YoutubeDL(opzioni) as ydl:
            ydl.download([row["trailer_link"]])
        scaricati_ora += 1
        print(f"  ok ({scaricati_ora}): {row['genere_principale']}")
    except Exception:
        pass

    if scaricati_ora >= 30:
        break
    time.sleep(2)

print("Scaricati in questo lotto:", scaricati_ora)
```
